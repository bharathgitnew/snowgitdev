```python
def get_cols(sql_text):
    cols = re.findall('VALUE:(c[0-9]+)::(\w+) AS (\w+)',sql_text)
    new_col_list = []
    
    for col in cols:
        if col[1] in ("STRING", "VARCHAR", "CHAR"):
            c_type = 'VARCHAR(16777216)'
        elif col[1] in ("NUMBER", "DECIMAL"):
            c_type = 'NUMBER(38,10)'
        elif col[1] in ("BOOLEAN"):
            c_type = 'BOOLEAN'
        elif col[1] in ("DATE", "TIMESTAMP_NTZ", "TIMESTAMP", "DATETIME"):
            c_type = 'TIMESTAMP_NTZ(9)'
        else:
            c_type = 'VARCHAR(16777216)'
        
        new_col_list.append([col[0], c_type, col[2]])
    return new_col_list
```

```python
def generate_source_block(table_name, schema_name, cols, primary_keys, incremental):
    source_block = """
    WITH source_data AS (
        SELECT
    """
    
    for col in cols:
        
        source_block += "\t\tVALUE:{c_num}::{c_type} AS {c_name}".format(c_num = col[0], c_type = col[1], c_name = col[2])
        if col == cols[-1]:
            source_block += ''
        else:
            source_block += ',\n'
    
    if incremental:
        source_block += """
        FROM {{{{ source("{schema_name}", "{table_name}")}}}}
        {{%- if is_incremental() %}}
            WHERE TO_DATE(_Filedate) = (SELECT TO_DATE(MAX(_FileDate)) FROM {{{{ source("{schema_name}", "{table_name}") }}}}) 
        {{%- else %}}
            QUALIFY (ROW_NUMBER() OVER ( PARTITION BY {primary_keys} ORDER BY _Filedate DESC )) = 1
        {{%- endif %}}
        ),
        """.format(table_name=table_name, schema_name=schema_name, primary_keys=primary_keys)
    else:
        source_block += """
        FROM {{{{ source("{schema_name}", "{table_name}")}}}}
        WHERE TO_DATE(_Filedate) = (SELECT TO_DATE(MAX(_FileDate)) FROM {{{{ source("{schema_name}", "{table_name}") }}}})
        AND TO_DATE(_UPLOADDATE) = (SELECT TO_DATE(MAX(_UPLOADDATE)) FROM {{{{ source("{schema_name}", "{table_name}") }}}})
        """.format(table_name=table_name, schema_name=schema_name)
    return source_block
            
    
    
        
```

```python
def generate_transform_block(cols):
    transform_block = """
    transform_data AS (
        SELECT
    """
    
    for col in cols:
        if col == cols[-1]:
            transform_block += """
            {c_name},
            '{{{{ invocation_id }}}}' AS AUDIT_LOG_ID,
            SHA2('hash_me') AS _row_hash
            """.format(c_name=col[2])
        else:
            transform_block += "\t\t{c_name},\n".format(c_name=col[2])
        
    transform_block += """
        FROM source_data
    ),
    """
        
    return transform_block

    
        
```

```python
def generate_view_source_block(cols, schema_name, table_name):
    source_block = """
    source_data AS (
        SELECT
    """
    
    for col in cols:
        if col == cols[-1]:
            source_block += "\t\t{c_name}\n".format(c_name=col[2])
        else:
            source_block += "\t\t{c_name},\n".format(c_name=col[2])
        
    source_block += """
        FROM {{{{ source('ODS_{schema_name}', '{table_name}') }}}}
    ),
    """.format(schema_name=schema_name, table_name=table_name)
        
    return source_block
```
