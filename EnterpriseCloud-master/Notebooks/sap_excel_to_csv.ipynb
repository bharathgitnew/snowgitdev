Conversion and clean up for .xlsx SAP Manual Loads to .csv
 

```python
import os
import pandas as pd
import shutil
from datetime import datetime
```

```python

folder_path = 'F:\Shares\Files\SAPH-API'

archive_folder = os.path.join(folder_path, 'archive_manual_loads')
if not os.path.exists(archive_folder):
    os.makedirs(archive_folder)

xlsx_files = [file for file in os.listdir(folder_path) if file.lower().endswith('.xlsx')]
timestamp = datetime.now().strftime("%Y%m%d%H%M%S")
# Iterate over each XLSX file
for xlsx_file in xlsx_files:
    print(f"Processing file: {xlsx_file}")
    file_path = os.path.join(folder_path, xlsx_file)
    
    table_name = os.path.splitext(xlsx_file)[0]
    
    
    # Convert XLSX file to CSV
    csv_file = f"{table_name}_{timestamp}.csv"
    csv_file_path = os.path.join(folder_path, csv_file)
    
    try:
        # Read XLSX data
        excel_data = pd.read_excel(file_path, dtype=str)
        print(excel_data.head())
        # Modify column names
        excel_data.rename(columns=lambda x: x.replace('/', '_'), inplace=True)
        # Convert all columns to string type
        excel_data = excel_data.astype(str)
        # Strip leading and trailing spaces from column names
        excel_data.columns = excel_data.columns.str.strip()
        # Replace 'nan' strings with blank values
        excel_data.replace({'nan': ''}, inplace=True)
        # Save modified data to CSV
        excel_data.to_csv(csv_file_path, index=False)
        
        print(f"Conversion successful. CSV file saved: {csv_file}")

        # Archive the XLSX file with timestamp
        archive_file = f"{table_name}_{timestamp}.xlsx"
        archive_file_path = os.path.join(folder_path, "archive_manual_loads", archive_file)
        shutil.move(file_path, archive_file_path)
        print(f"XLSX file archived: {xlsx_file}\n")
        
    except Exception as e:
        print(f"Error processing file: {xlsx_file}")
        print(f"Error message: {str(e)}")
    
# Compare XLSX and CSV file count
xlsx_count = len(xlsx_files)
csv_count = len([file for file in os.listdir(folder_path) if file.lower().endswith('.csv')])

if xlsx_count == csv_count:
    print("XLSX and CSV file counts match.")
else:
    print("XLSX and CSV file counts do not match.")
    print(f"Number of XLSX files: {xlsx_count}")
    print(f"Number of CSV files: {csv_count}")
```
