```python
import json
import glob
import os
import sys
import datetime
import logging
import requests
```

```python
USERNAME = 'TELb2bsupport'
SECRET = os.environ['SIRFIS_API_SECRET']
print(SECRET)
API_URL =	'https://supplier.intel.com/SIRFIS/feeds/forecasts'
```

```python
# helper functions
def get_current_timestamp():
	logging.debug("Getting current timestamp")
	CURRENT_TIMESTAMP = str(datetime.datetime.now()).replace(" ","_").replace(":","_").replace(".","_")

	return CURRENT_TIMESTAMP

def get_sirfis_data(session, url, include_history=False, group_children=False, headers={}):

	response = session.get(url, headers=headers, data={"includeHistory": include_history, "groupChildren": group_children})

	status_code = response.status_code

	if status_code == 200:
		sirifis_data = json.loads(response.content)
	else:
		logging.error("API Error status code: {}".format(status_code))
	

	return sirifis_data

def write_results(data,timestamp):
	logging.debug("Writing to file...")
	with open('\\\\TELUSDWAPPPROD\Shares\Files\SIRFIS\SIRFIS_FORECAST_{}.json'.format(timestamp), 'w') as json_file:
		json.dump(data, json_file)
	logging.debug("Wrote to file {}".format(json_file))

	return json_file

def InvokeAPI():
	
	session = requests.Session()
	session.auth = (USERNAME, SECRET)

	session.cookies.set("SecureCHALLENGE","YES")

	headers = {'Accept': 'text/html,application/xhtml+xml,application/xml;q=0.9,image/avif,image/webp,image/apng,*/*;q=0.8,application/signed-exchange;v=b3;q=0.9',
		'Connection': 'keep-alive', 'User-Agent': "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.106 Safari/537.36",
		'Host': 'supplier.intel.com', 'Sec-Fetch-Dest': 'document', 'Sec-Fetch-Mode': 'navigate', 'Sec-Fetch-Site': 'none', 'Sec-Fetch-User': '?1',
		'Upgrade-Insecure-Requests': "1", 'Accept-Encoding': "gzip, deflate, br", 'Accept-Language': "en-US,en;q=0.9"}

	data = get_sirfis_data(session, API_URL, headers=headers)

	current_timestamp = get_current_timestamp()
	# print(data)
	output_file = write_results(data,current_timestamp)
	# print(output_file)
```

```python
InvokeAPI()

```
