```python
import pandas as pd
import subprocess 
```

```python
df = pd.read_excel(r'C:\Repo\EnterpriseCloud\.github\ntbks\git_env_vars.xlsx', usecols=['Key', 'Value'])
print(df.head())
```

```python
# #so i have a github cli command like gh variable set SNOWFLAKE_TEA_TECHNICAL_TRAINING_WH --env dev --body 'TEA_TECHNICAL_TRAINING_DEV_WH' i need to iterate oover the df and create this command for Key and Value pairs and print the command for each pair
# for index, row in df.iterrows():
#     key = row['Key']
#     value = row['Value']
#     command = f"gh variable set {key} --env SIT --body '{value}'"
#     print(command)


  
for index, row in df.iterrows():
    key = row['Key']
    value = row['Value']
    command = f"gh variable set {key} --env SIT --body {value}"
    print(f"Running command: {command}")
    result = subprocess.run(command, shell=True, capture_output=True, text=True)
    if result.returncode == 0:
        print(f"Command succeeded: {command}")
        print(f"Output:\n{result.stdout}")
    else:
        print(f"Command failed: {command}")
        print(f"Error:\n{result.stderr}")
```
